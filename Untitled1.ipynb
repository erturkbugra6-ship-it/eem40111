```python

import pandas as pd

url = "https://physionet.org/files/bidmc/1.0.0/bidmc_csv/bidmc_01_Signals.csv"

df = pd.read_csv(url)

print("Sütun adları:")
print(df.columns.tolist())

print("\nİlk 5 satır:")
print(df.head())

time_col = next(
    (c for c in df.columns
     if c.strip().lower() in ["time", "time (s)", "time_s"]),
    None
)

if time_col is not None:
    t = pd.to_numeric(df[time_col], errors="coerce").dropna()
    dt = t.diff().dropna()
    dt = dt[dt > 0]

    if len(dt) > 0:
        fs = 1 / dt.median()
        print(f"\nZaman sütunu: {time_col}")
        print(f"Örnekleme frekansı: {fs:.6f} Hz")
    else:
        print("Örnekleme frekansı hesaplanamadı.")
else:
    print("Zaman sütunu bulunamadı.")
```

```python
# 1. Zaman sütununu saniye cinsinden al.
t = pd.to_numeric(df["Time [s]"], errors="coerce").dropna()

# 2. Ardışık zaman örnekleri arasındaki farkı hesapla.
dt = t.diff().dropna()

# 3. Sıfır veya negatif farkları çıkar.
dt = dt[dt > 0]

# 4. Medyan zaman aralığının tersinden örnekleme frekansını hesapla.
fs = 1 / dt.median()

# 5. Sonucu ekrana yazdır.
print(f"Zaman sütunu: Time [s]")
print(f"Örnekleme frekansı: {fs:.6f} Hz")
print(f"Örnek sayısı: {len(df)}")
```

```python
import numpy as np
import pandas as pd
from scipy.signal import find_peaks

pencere_ornek = int(round(10 * fs))
pencere_sayisi = len(ppg) // pencere_ornek
satirlar = []

for i in range(pencere_sayisi):
    bas = i * pencere_ornek
    son = bas + pencere_ornek

    ppg_parca = ppg[bas:son]
    ekg_parca = ekg[bas:son]

    p_tepeler, _ = find_peaks(
        ppg_parca,
        distance=int(np.ceil(0.30 * fs)),
        prominence=0.05
    )
    r_tepeler, _ = find_peaks(
        ekg_parca,
        distance=int(np.ceil(0.30 * fs)),
        prominence=0.50
    )

    p_nabiz = np.nan
    e_nabiz = np.nan

    if len(p_tepeler) >= 2:
        p_nabiz = 60 / np.mean(np.diff(p_tepeler) / fs)
    if len(r_tepeler) >= 2:
        e_nabiz = 60 / np.mean(np.diff(r_tepeler) / fs)

    satirlar.append({
        "Başlangıç (s)": bas / fs,
        "PPG tepe sayısı": len(p_tepeler),
        "EKG tepe sayısı": len(r_tepeler),
        "PPG nabzı": p_nabiz,
        "EKG nabzı": e_nabiz,
        "Mutlak hata": abs(p_nabiz - e_nabiz)
    })

baseline = pd.DataFrame(satirlar)
gecerli = baseline["Mutlak hata"].notna()

print(f"Toplam pencere: {len(baseline)}")
print(f"İki nabzın da hesaplandığı pencere: {gecerli.sum()}")
print(f"Hesaplanamayan pencere: {(~gecerli).sum()}")

if gecerli.any():
    baseline_mae = baseline.loc[gecerli, "Mutlak hata"].mean()
    print(f"Gürültüsüz ham PPG baseline MAE: {baseline_mae:.4f} atım/dk")
    print("\nMutlak hatası en büyük 5 pencere:")
    print(baseline.loc[gecerli].nlargest(5, "Mutlak hata")
          .to_string(index=False, float_format="%.4f"))
else:
    print("Baseline MAE: ___")
```

```python
rng_kayit = np.random.default_rng(4011)
n_kullan = pencere_sayisi * pencere_ornek
ppg_asıl = ppg[:n_kullan]
gurultu_kayit = np.zeros(n_kullan)

for i in range(pencere_sayisi):
    a = i * pencere_ornek
    b = a + pencere_ornek
    x = ppg_asıl[a:b]

    guc = np.mean((x - x.mean()) ** 2)
    if guc <= 0:
        raise ValueError(f"{i}. pencerede sinyal gücü sıfır.")

    z = rng_kayit.normal(size=len(x))
    z = z - z.mean()
    z = z * np.sqrt((guc / 100) / np.mean(z ** 2))
    gurultu_kayit[a:b] = z

ppg_kayit_20db = ppg_asıl + gurultu_kayit

yontemler_kayit = {
    "Ham tepe": ppg_kayit_20db,
    "Hareketli ortalama": uniform_filter1d(
        ppg_kayit_20db, size=ma_ornek, mode="reflect"
    ),
    "Butterworth": sosfiltfilt(sos, ppg_kayit_20db)
}

deney_20db = baseline[["Başlangıç (s)", "EKG nabzı"]].copy()

for ad, sinyal in yontemler_kayit.items():
    tahminler = []

    for i in range(pencere_sayisi):
        a = i * pencere_ornek
        b = a + pencere_ornek
        pk, _ = find_peaks(
            sinyal[a:b],
            distance=int(np.ceil(0.30 * fs)),
            prominence=0.05
        )

        tahmin = np.nan
        if len(pk) >= 2:
            tahmin = 60 / np.mean(np.diff(pk) / fs)

        tahminler.append(tahmin)

    deney_20db[ad] = tahminler

adlar = list(yontemler_kayit)
ortak_20db = deney_20db[["EKG nabzı"] + adlar].notna().all(axis=1)

print(f"20 dB — ortak pencere: {ortak_20db.sum()} / {pencere_sayisi}")

for ad in adlar:
    print(f"{ad} — hesaplanamayan pencere: {deney_20db[ad].isna().sum()}")

    if ortak_20db.any():
        hata = abs(
            deney_20db.loc[ortak_20db, ad]
            - deney_20db.loc[ortak_20db, "EKG nabzı"]
        )
        print(f"{ad} — MAE: {hata.mean():.4f} atım/dk")
    else:
        print(f"{ad} — MAE: ___")
```

```python
snr_duzeyleri = [20, 10, 0]
tekrar_sayisi = 5
deney_satirlari = []

x_pencereler = ppg_asıl.reshape(pencere_sayisi, pencere_ornek)
pencere_gucleri = np.mean(
    (x_pencereler - x_pencereler.mean(axis=1, keepdims=True)) ** 2,
    axis=1,
    keepdims=True
)
ekg_referans = baseline["EKG nabzı"].to_numpy()

for tekrar in range(tekrar_sayisi):
    rng_deney = np.random.default_rng(4011 + tekrar)
    z = rng_deney.normal(size=x_pencereler.shape)
    z -= z.mean(axis=1, keepdims=True)
    z /= np.sqrt(np.mean(z ** 2, axis=1, keepdims=True))

    for snr_db in snr_duzeyleri:
        gurultu = z * np.sqrt(
            pencere_gucleri / (10 ** (snr_db / 10))
        )
        x_gurultulu = (x_pencereler + gurultu).ravel()

        sinyaller = {
            "Ham tepe": x_gurultulu,
            "Hareketli ortalama": uniform_filter1d(
                x_gurultulu, size=ma_ornek, mode="reflect"
            ),
            "Butterworth": sosfiltfilt(sos, x_gurultulu)
        }

        tahminler = {}
        for ad, sinyal in sinyaller.items():
            nabizlar = np.full(pencere_sayisi, np.nan)

            for i in range(pencere_sayisi):
                a = i * pencere_ornek
                b = a + pencere_ornek
                pk, _ = find_peaks(
                    sinyal[a:b],
                    distance=int(np.ceil(0.30 * fs)),
                    prominence=0.05
                )
                if len(pk) >= 2:
                    nabizlar[i] = 60 / np.mean(np.diff(pk) / fs)

            tahminler[ad] = nabizlar

        ortak = np.isfinite(ekg_referans)
        for nabizlar in tahminler.values():
            ortak = ortak & np.isfinite(nabizlar)

        for ad, nabizlar in tahminler.items():
            mae = np.nan
            if ortak.any():
                mae = np.mean(abs(
                    nabizlar[ortak] - ekg_referans[ortak]
                ))

            deney_satirlari.append({
                "SNR (dB)": snr_db,
                "Tekrar": tekrar + 1,
                "Yöntem": ad,
                "MAE": mae,
                "Ortak pencere": int(ortak.sum()),
                "Hesaplanamayan": int(np.isnan(nabizlar).sum())
            })

deney_sonuclari = pd.DataFrame(deney_satirlari)

ozet = deney_sonuclari.groupby(
    ["SNR (dB)", "Yöntem"]
).agg(
    Ortalama_MAE=("MAE", "mean"),
    MAE_std=("MAE", "std"),
    En_az_ortak_pencere=("Ortak pencere", "min"),
    Toplam_hesaplanamayan=("Hesaplanamayan", "sum")
).reset_index().sort_values(
    ["SNR (dB)", "Yöntem"], ascending=[False, True]
)

print(ozet.to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
import pandas as pd
import numpy as np

url_02 = (
    "https://physionet.org/files/bidmc/1.0.0/"
    "bidmc_csv/bidmc_02_Signals.csv"
)
df_02 = pd.read_csv(url_02)
df_02.columns = df_02.columns.str.strip()

print("Tablo boyutu:", df_02.shape)
print("Sütunlar:", df_02.columns.tolist())
print("\nİlk 5 satır:")
print(df_02.head())

zaman_02_dosya = df_02["Time [s]"].to_numpy(dtype=float)
fs_02_aday = (
    (len(zaman_02_dosya) - 1)
    / (zaman_02_dosya[-1] - zaman_02_dosya[0])
)
zaman_02_aday = (
    zaman_02_dosya[0] + np.arange(len(df_02)) / fs_02_aday
)

print(f"\nSüre ve örnek sayısından aday frekans: {fs_02_aday:.6f} Hz")
print("Zamanlar sonlu mu?:", np.isfinite(zaman_02_dosya).all())
print("Zamanın geriye gittiği aralık sayısı:",
      np.sum(np.diff(zaman_02_dosya) < 0))
print("Tekrarlanan zaman aralığı sayısı:",
      np.sum(np.diff(zaman_02_dosya) == 0))
print("Düzenli zaman dizisinden en büyük mutlak sapma:",
      np.max(np.abs(zaman_02_dosya - zaman_02_aday)), "saniye")
```

```python
import numpy as np
import matplotlib.pyplot as plt

fs_02 = float(fs_02_aday)
zaman_02 = np.arange(len(df_02)) / fs_02
ppg_02 = df_02["PLETH"].to_numpy(dtype=float)
ekg_02 = df_02["II"].to_numpy(dtype=float)

maske_02 = zaman_02 < 10

fig, eksenler = plt.subplots(2, 1, figsize=(14, 7), sharex=True)

eksenler[0].plot(zaman_02[maske_02], ppg_02[maske_02])
eksenler[0].set_ylabel("PPG genliği")
eksenler[0].set_title("bidmc_02 — İlk 10 saniye")

eksenler[1].plot(
    zaman_02[maske_02], ekg_02[maske_02], color="purple"
)
eksenler[1].set_ylabel("EKG — II")
eksenler[1].set_xlabel("Zaman (s)")

for eksen in eksenler:
    eksen.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
from scipy.signal import find_peaks

t02_10 = zaman_02[maske_02]
p02_10 = ppg_02[maske_02]
e02_10 = ekg_02[maske_02]

p02_pk, _ = find_peaks(
    p02_10,
    distance=int(np.ceil(0.30 * fs_02)),
    prominence=0.05
)
r02_pk, _ = find_peaks(
    e02_10,
    distance=int(np.ceil(0.30 * fs_02)),
    prominence=0.25
)

fig, eksenler = plt.subplots(2, 1, figsize=(14, 7), sharex=True)

for eksen, sinyal, tepeler, ad in [
    (eksenler[0], p02_10, p02_pk, "Ham PPG"),
    (eksenler[1], e02_10, r02_pk, "EKG — II")
]:
    eksen.plot(t02_10, sinyal, label=ad)
    eksen.plot(t02_10[tepeler], sinyal[tepeler], "rx",
               label="Tepe adayları")
    eksen.set_ylabel(ad)
    eksen.grid(True, alpha=0.3)
    eksen.legend()

eksenler[0].set_title("bidmc_02 — İlk 10 saniyede tepe kontrolü")
eksenler[1].set_xlabel("Zaman (s)")
plt.tight_layout()
plt.show()

print("PPG tepe sayısı:", len(p02_pk))
print("EKG R adayı sayısı:", len(r02_pk))
```

```python
if len(p02_pk) >= 2 and len(r02_pk) >= 2:
    nabiz_ppg_02 = 60 / np.mean(np.diff(p02_pk) / fs_02)
    nabiz_ekg_02 = 60 / np.mean(np.diff(r02_pk) / fs_02)
    hata_02 = abs(nabiz_ppg_02 - nabiz_ekg_02)

    print(f"Ham PPG nabzı: {nabiz_ppg_02:.4f} atım/dk")
    print(f"EKG referans nabzı: {nabiz_ekg_02:.4f} atım/dk")
    print(f"İlk 10 saniyenin mutlak hatası: {hata_02:.4f} atım/dk")
else:
    print("Nabız ve mutlak hata: ___ (yeterli tepe yok)")
```

```python
from scipy.signal import butter, sosfiltfilt, find_peaks
from scipy.ndimage import uniform_filter1d

pencere_02 = int(round(10 * fs_02))
pencere_sayisi_02 = len(ppg_02) // pencere_02

ma_ornek_02 = int(round(0.10 * fs_02))
if ma_ornek_02 % 2 == 0:
    ma_ornek_02 += 1

sos_02 = butter(
    4, [0.5, 8], btype="bandpass", fs=fs_02, output="sos"
)

sinyaller_02 = {
    "Ham tepe": ppg_02,
    "Hareketli ortalama": uniform_filter1d(
        ppg_02, size=ma_ornek_02, mode="reflect"
    ),
    "Butterworth": sosfiltfilt(sos_02, ppg_02)
}

satirlar_02 = []

for i in range(pencere_sayisi_02):
    a = i * pencere_02
    b = a + pencere_02

    r_pk, _ = find_peaks(
        ekg_02[a:b],
        distance=int(np.ceil(0.30 * fs_02)),
        prominence=0.25
    )

    ekg_nabiz = np.nan
    if len(r_pk) >= 2:
        ekg_nabiz = 60 / np.mean(np.diff(r_pk) / fs_02)

    satir = {"Başlangıç (s)": a / fs_02, "EKG nabzı": ekg_nabiz}

    for ad, sinyal in sinyaller_02.items():
        p_pk, _ = find_peaks(
            sinyal[a:b],
            distance=int(np.ceil(0.30 * fs_02)),
            prominence=0.05
        )

        nabiz = np.nan
        if len(p_pk) >= 2:
            nabiz = 60 / np.mean(np.diff(p_pk) / fs_02)

        satir[ad] = nabiz

    satirlar_02.append(satir)

baseline_02 = pd.DataFrame(satirlar_02)
adlar_02 = list(sinyaller_02)
ortak_02 = baseline_02[["EKG nabzı"] + adlar_02].notna().all(axis=1)

print(f"Ortak hesaplanabilen pencere: "
      f"{ortak_02.sum()} / {pencere_sayisi_02}")

for ad in adlar_02:
    baseline_02[ad + " hata"] = abs(
        baseline_02[ad] - baseline_02["EKG nabzı"]
    )
    print(f"{ad} — hesaplanamayan pencere: "
          f"{baseline_02[ad].isna().sum()}")

    if ortak_02.any():
        mae = baseline_02.loc[ortak_02, ad + " hata"].mean()
        print(f"{ad} — MAE: {mae:.4f} atım/dk")
    else:
        print(f"{ad} — MAE: ___")

print("\nHam yöntemde en yüksek hatalı 5 pencere:")
print(baseline_02.nlargest(5, "Ham tepe hata").to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
n_02 = pencere_sayisi_02 * pencere_02
x_02 = ppg_02[:n_02].reshape(pencere_sayisi_02, pencere_02)

guc_02 = np.mean(
    (x_02 - x_02.mean(axis=1, keepdims=True)) ** 2,
    axis=1, keepdims=True
)
if np.any(guc_02 <= 0):
    raise ValueError("En az bir pencerenin sinyal gücü sıfır.")

referans_02 = baseline_02["EKG nabzı"].to_numpy()
deney_satirlari_02 = []

for tekrar in range(5):
    rng_02 = np.random.default_rng(4021 + tekrar)
    z_02 = rng_02.normal(size=x_02.shape)
    z_02 -= z_02.mean(axis=1, keepdims=True)
    z_02 /= np.sqrt(np.mean(z_02 ** 2, axis=1, keepdims=True))

    for snr_db in [20, 10, 0]:
        gurultu_02 = z_02 * np.sqrt(
            guc_02 / (10 ** (snr_db / 10))
        )
        x_gurultulu_02 = (x_02 + gurultu_02).ravel()

        sinyaller_deney_02 = {
            "Ham tepe": x_gurultulu_02,
            "Hareketli ortalama": uniform_filter1d(
                x_gurultulu_02, size=ma_ornek_02, mode="reflect"
            ),
            "Butterworth": sosfiltfilt(sos_02, x_gurultulu_02)
        }

        tahminler_02 = {}
        for ad, sinyal in sinyaller_deney_02.items():
            nabizlar = np.full(pencere_sayisi_02, np.nan)

            for i in range(pencere_sayisi_02):
                a = i * pencere_02
                b = a + pencere_02
                pk, _ = find_peaks(
                    sinyal[a:b],
                    distance=int(np.ceil(0.30 * fs_02)),
                    prominence=0.05
                )
                if len(pk) >= 2:
                    nabizlar[i] = 60 / np.mean(np.diff(pk) / fs_02)

            tahminler_02[ad] = nabizlar

        ortak = np.isfinite(referans_02)
        for nabizlar in tahminler_02.values():
            ortak = ortak & np.isfinite(nabizlar)

        for ad, nabizlar in tahminler_02.items():
            mae = np.nan
            if ortak.any():
                mae = np.mean(abs(
                    nabizlar[ortak] - referans_02[ortak]
                ))

            deney_satirlari_02.append({
                "Kayıt": "bidmc_02",
                "SNR (dB)": snr_db,
                "Tekrar": tekrar + 1,
                "Yöntem": ad,
                "MAE": mae,
                "Ortak pencere": int(ortak.sum()),
                "Hesaplanamayan": int(np.isnan(nabizlar).sum())
            })

deney_sonuclari_02 = pd.DataFrame(deney_satirlari_02)

ozet_02 = deney_sonuclari_02.groupby(
    ["SNR (dB)", "Yöntem"]
).agg(
    Ortalama_MAE=("MAE", "mean"),
    MAE_std=("MAE", "std"),
    En_az_ortak_pencere=("Ortak pencere", "min"),
    Toplam_hesaplanamayan=("Hesaplanamayan", "sum")
).reset_index().sort_values(
    ["SNR (dB)", "Yöntem"], ascending=[False, True]
)

deney_sonuclari_02.to_csv("bidmc_02_gurultu_tekrarlari.csv", index=False)
ozet_02.to_csv("bidmc_02_SNR_ozet.csv", index=False)

print(ozet_02.to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
from scipy.signal import butter, sosfiltfilt, find_peaks
from scipy.ndimage import uniform_filter1d

pencere_03 = int(round(10 * fs_03))
pencere_sayisi_03 = len(ppg_03) // pencere_03

ma_ornek_03 = int(round(0.10 * fs_03))
if ma_ornek_03 % 2 == 0:
    ma_ornek_03 += 1

sos_03 = butter(
    4, [0.5, 8], btype="bandpass", fs=fs_03, output="sos"
)

sinyaller_03 = {
    "Ham tepe": ppg_03,
    "Hareketli ortalama": uniform_filter1d(
        ppg_03, size=ma_ornek_03, mode="reflect"
    ),
    "Butterworth": sosfiltfilt(sos_03, ppg_03)
}

satirlar_03 = []

for i in range(pencere_sayisi_03):
    a = i * pencere_03
    b = a + pencere_03

    r_pk, _ = find_peaks(
        ekg_03[a:b],
        distance=int(np.ceil(0.30 * fs_03)),
        prominence=0.10
    )

    ekg_nabiz = np.nan
    if len(r_pk) >= 2:
        ekg_nabiz = 60 / np.mean(np.diff(r_pk) / fs_03)

    satir = {
        "Başlangıç (s)": a / fs_03,
        "EKG R adayı sayısı": len(r_pk),
        "EKG nabzı": ekg_nabiz
    }

    for ad, sinyal in sinyaller_03.items():
        p_pk, _ = find_peaks(
            sinyal[a:b],
            distance=int(np.ceil(0.30 * fs_03)),
            prominence=0.05
        )

        nabiz = np.nan
        if len(p_pk) >= 2:
            nabiz = 60 / np.mean(np.diff(p_pk) / fs_03)

        satir[ad] = nabiz
        satir[ad + " tepe sayısı"] = len(p_pk)

    satirlar_03.append(satir)

baseline_03 = pd.DataFrame(satirlar_03)
adlar_03 = list(sinyaller_03)
ortak_03 = baseline_03[["EKG nabzı"] + adlar_03].notna().all(axis=1)

print(f"Ortak hesaplanabilen pencere: "
      f"{ortak_03.sum()} / {pencere_sayisi_03}")
print("EKG hesaplanamayan pencere:",
      baseline_03["EKG nabzı"].isna().sum())

for ad in adlar_03:
    baseline_03[ad + " hata"] = abs(
        baseline_03[ad] - baseline_03["EKG nabzı"]
    )
    print(f"{ad} — hesaplanamayan pencere: "
          f"{baseline_03[ad].isna().sum()}")

    if ortak_03.any():
        mae = baseline_03.loc[ortak_03, ad + " hata"].mean()
        print(f"{ad} — MAE: {mae:.4f} atım/dk")
    else:
        print(f"{ad} — MAE: ___")

print("\nHam yöntemde en yüksek hatalı 5 pencere:")
print(baseline_03.nlargest(5, "Ham tepe hata").to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
n_03 = pencere_sayisi_03 * pencere_03
x_03 = ppg_03[:n_03].reshape(pencere_sayisi_03, pencere_03)

guc_03 = np.mean(
    (x_03 - x_03.mean(axis=1, keepdims=True)) ** 2,
    axis=1, keepdims=True
)
if np.any(guc_03 <= 0):
    raise ValueError("En az bir pencerenin sinyal gücü sıfır.")

referans_03 = baseline_03_v["V referans nabzı"].to_numpy()
deney_satirlari_03 = []

for tekrar in range(5):
    rng_03 = np.random.default_rng(4031 + tekrar)
    z_03 = rng_03.normal(size=x_03.shape)
    z_03 -= z_03.mean(axis=1, keepdims=True)
    z_03 /= np.sqrt(np.mean(z_03 ** 2, axis=1, keepdims=True))

    for snr_db in [20, 10, 0]:
        gurultu_03 = z_03 * np.sqrt(
            guc_03 / (10 ** (snr_db / 10))
        )
        x_gurultulu_03 = (x_03 + gurultu_03).ravel()

        sinyaller_deney_03 = {
            "Ham tepe": x_gurultulu_03,
            "Hareketli ortalama": uniform_filter1d(
                x_gurultulu_03, size=ma_ornek_03, mode="reflect"
            ),
            "Butterworth": sosfiltfilt(sos_03, x_gurultulu_03)
        }

        tahminler_03 = {}
        for ad, sinyal in sinyaller_deney_03.items():
            nabizlar = np.full(pencere_sayisi_03, np.nan)

            for i in range(pencere_sayisi_03):
                a = i * pencere_03
                b = a + pencere_03
                pk, _ = find_peaks(
                    sinyal[a:b],
                    distance=int(np.ceil(0.30 * fs_03)),
                    prominence=0.05
                )
                if len(pk) >= 2:
                    nabizlar[i] = 60 / np.mean(np.diff(pk) / fs_03)

            tahminler_03[ad] = nabizlar

        ortak = np.isfinite(referans_03)
        for nabizlar in tahminler_03.values():
            ortak = ortak & np.isfinite(nabizlar)

        for ad, nabizlar in tahminler_03.items():
            mae = np.nan
            if ortak.any():
                mae = np.mean(abs(
                    nabizlar[ortak] - referans_03[ortak]
                ))

            deney_satirlari_03.append({
                "Kayıt": "bidmc_03",
                "EKG referansı": "V — geçici",
                "SNR (dB)": snr_db,
                "Tekrar": tekrar + 1,
                "Yöntem": ad,
                "MAE": mae,
                "Ortak pencere": int(ortak.sum()),
                "Hesaplanamayan": int(np.isnan(nabizlar).sum())
            })

deney_sonuclari_03 = pd.DataFrame(deney_satirlari_03)

ozet_03 = deney_sonuclari_03.groupby(
    ["SNR (dB)", "Yöntem"]
).agg(
    Ortalama_MAE=("MAE", "mean"),
    MAE_std=("MAE", "std"),
    En_az_ortak_pencere=("Ortak pencere", "min"),
    Toplam_hesaplanamayan=("Hesaplanamayan", "sum")
).reset_index().sort_values(
    ["SNR (dB)", "Yöntem"], ascending=[False, True]
)

deney_sonuclari_03.to_csv("bidmc_03_gurultu_tekrarlari.csv", index=False)
ozet_03.to_csv("bidmc_03_SNR_ozet.csv", index=False)

print(ozet_03.to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
from scipy.signal import butter, sosfiltfilt, find_peaks
from scipy.ndimage import uniform_filter1d

pencere_04 = int(round(10 * fs_04))
pencere_sayisi_04 = len(ppg_04) // pencere_04

ma_ornek_04 = int(round(0.10 * fs_04))
if ma_ornek_04 % 2 == 0:
    ma_ornek_04 += 1

sos_04 = butter(
    4, [0.5, 8], btype="bandpass", fs=fs_04, output="sos"
)

sinyaller_04 = {
    "Ham tepe": ppg_04,
    "Hareketli ortalama": uniform_filter1d(
        ppg_04, size=ma_ornek_04, mode="reflect"
    ),
    "Butterworth": sosfiltfilt(sos_04, ppg_04)
}

satirlar_04 = []

for i in range(pencere_sayisi_04):
    a = i * pencere_04
    b = a + pencere_04

    v_pk, _ = find_peaks(
        -v04[a:b],
        distance=int(np.ceil(0.30 * fs_04)),
        prominence=0.20
    )

    v_nabiz = np.nan
    if len(v_pk) >= 2:
        v_nabiz = 60 / np.mean(np.diff(v_pk) / fs_04)

    satir = {
        "Başlangıç (s)": a / fs_04,
        "V aday sayısı": len(v_pk),
        "V referans nabzı": v_nabiz
    }

    for ad, sinyal in sinyaller_04.items():
        pk, _ = find_peaks(
            sinyal[a:b],
            distance=int(np.ceil(0.30 * fs_04)),
            prominence=0.05
        )

        nabiz = np.nan
        if len(pk) >= 2:
            nabiz = 60 / np.mean(np.diff(pk) / fs_04)

        satir[ad] = nabiz
        satir[ad + " tepe sayısı"] = len(pk)

    satirlar_04.append(satir)

baseline_04 = pd.DataFrame(satirlar_04)
adlar_04 = list(sinyaller_04)
ortak_04 = baseline_04[
    ["V referans nabzı"] + adlar_04
].notna().all(axis=1)

print(f"Ortak hesaplanabilen pencere: "
      f"{ortak_04.sum()} / {pencere_sayisi_04}")
print("V referansı hesaplanamayan pencere:",
      baseline_04["V referans nabzı"].isna().sum())

for ad in adlar_04:
    baseline_04[ad + " hata"] = abs(
        baseline_04[ad] - baseline_04["V referans nabzı"]
    )
    print(f"{ad} — hesaplanamayan pencere: "
          f"{baseline_04[ad].isna().sum()}")

    if ortak_04.any():
        mae = baseline_04.loc[ortak_04, ad + " hata"].mean()
        print(f"{ad} — ön MAE: {mae:.4f} atım/dk")
    else:
        print(f"{ad} — ön MAE: ___")

print("\nHam yöntemde en yüksek hatalı 5 pencere:")
print(baseline_04.nlargest(5, "Ham tepe hata").to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
n_04 = pencere_sayisi_04 * pencere_04
x_04 = ppg_04[:n_04].reshape(pencere_sayisi_04, pencere_04)

guc_04 = np.mean(
    (x_04 - x_04.mean(axis=1, keepdims=True)) ** 2,
    axis=1, keepdims=True
)
if np.any(guc_04 <= 0):
    raise ValueError("En az bir pencerenin sinyal gücü sıfır.")

referans_04 = baseline_04["V referans nabzı"].to_numpy()
deney_satirlari_04 = []

for tekrar in range(5):
    rng_04 = np.random.default_rng(4041 + tekrar)
    z_04 = rng_04.normal(size=x_04.shape)
    z_04 -= z_04.mean(axis=1, keepdims=True)
    z_04 /= np.sqrt(np.mean(z_04 ** 2, axis=1, keepdims=True))

    for snr_db in [20, 10, 0]:
        gurultu_04 = z_04 * np.sqrt(
            guc_04 / (10 ** (snr_db / 10))
        )
        x_gurultulu_04 = (x_04 + gurultu_04).ravel()

        sinyaller_deney_04 = {
            "Ham tepe": x_gurultulu_04,
            "Hareketli ortalama": uniform_filter1d(
                x_gurultulu_04, size=ma_ornek_04, mode="reflect"
            ),
            "Butterworth": sosfiltfilt(sos_04, x_gurultulu_04)
        }

        tahminler_04 = {}
        for ad, sinyal in sinyaller_deney_04.items():
            nabizlar = np.full(pencere_sayisi_04, np.nan)

            for i in range(pencere_sayisi_04):
                a = i * pencere_04
                b = a + pencere_04
                pk, _ = find_peaks(
                    sinyal[a:b],
                    distance=int(np.ceil(0.30 * fs_04)),
                    prominence=0.05
                )
                if len(pk) >= 2:
                    nabizlar[i] = 60 / np.mean(np.diff(pk) / fs_04)

            tahminler_04[ad] = nabizlar

        ortak = np.isfinite(referans_04)
        for nabizlar in tahminler_04.values():
            ortak = ortak & np.isfinite(nabizlar)

        for ad, nabizlar in tahminler_04.items():
            mae = np.nan
            if ortak.any():
                mae = np.mean(abs(
                    nabizlar[ortak] - referans_04[ortak]
                ))

            deney_satirlari_04.append({
                "Kayıt": "bidmc_04",
                "EKG referansı": "V — geçici",
                "SNR (dB)": snr_db,
                "Tekrar": tekrar + 1,
                "Yöntem": ad,
                "MAE": mae,
                "Ortak pencere": int(ortak.sum()),
                "Hesaplanamayan": int(np.isnan(nabizlar).sum())
            })

deney_sonuclari_04 = pd.DataFrame(deney_satirlari_04)

ozet_04 = deney_sonuclari_04.groupby(
    ["SNR (dB)", "Yöntem"]
).agg(
    Ortalama_MAE=("MAE", "mean"),
    MAE_std=("MAE", "std"),
    En_az_ortak_pencere=("Ortak pencere", "min"),
    Toplam_hesaplanamayan=("Hesaplanamayan", "sum")
).reset_index().sort_values(
    ["SNR (dB)", "Yöntem"], ascending=[False, True]
)

deney_sonuclari_04.to_csv("bidmc_04_gurultu_tekrarlari.csv", index=False)
ozet_04.to_csv("bidmc_04_SNR_ozet.csv", index=False)

print(ozet_04.to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
from scipy.signal import butter, sosfiltfilt, find_peaks
from scipy.ndimage import uniform_filter1d

pencere_05 = int(round(10 * fs_05))
pencere_sayisi_05 = len(ppg_05) // pencere_05

ma_ornek_05 = int(round(0.10 * fs_05))
if ma_ornek_05 % 2 == 0:
    ma_ornek_05 += 1

sos_05 = butter(
    4, [0.5, 8], btype="bandpass", fs=fs_05, output="sos"
)

sinyaller_05 = {
    "Ham tepe": ppg_05,
    "Hareketli ortalama": uniform_filter1d(
        ppg_05, size=ma_ornek_05, mode="reflect"
    ),
    "Butterworth": sosfiltfilt(sos_05, ppg_05)
}

satirlar_05 = []

for i in range(pencere_sayisi_05):
    a = i * pencere_05
    b = a + pencere_05

    r_pk, _ = find_peaks(
        ekg_05[a:b],
        distance=int(np.ceil(0.30 * fs_05)),
        prominence=0.40
    )

    ekg_nabiz = np.nan
    if len(r_pk) >= 2:
        ekg_nabiz = 60 / np.mean(np.diff(r_pk) / fs_05)

    satir = {
        "Başlangıç (s)": a / fs_05,
        "EKG R adayı sayısı": len(r_pk),
        "EKG nabzı": ekg_nabiz
    }

    for ad, sinyal in sinyaller_05.items():
        pk, _ = find_peaks(
            sinyal[a:b],
            distance=int(np.ceil(0.30 * fs_05)),
            prominence=0.05
        )

        nabiz = np.nan
        if len(pk) >= 2:
            nabiz = 60 / np.mean(np.diff(pk) / fs_05)

        satir[ad] = nabiz
        satir[ad + " tepe sayısı"] = len(pk)

    satirlar_05.append(satir)

baseline_05 = pd.DataFrame(satirlar_05)
adlar_05 = list(sinyaller_05)
ortak_05 = baseline_05[
    ["EKG nabzı"] + adlar_05
].notna().all(axis=1)

print(f"Ortak hesaplanabilen pencere: "
      f"{ortak_05.sum()} / {pencere_sayisi_05}")
print("EKG hesaplanamayan pencere:",
      baseline_05["EKG nabzı"].isna().sum())

for ad in adlar_05:
    baseline_05[ad + " hata"] = abs(
        baseline_05[ad] - baseline_05["EKG nabzı"]
    )
    print(f"{ad} — hesaplanamayan pencere: "
          f"{baseline_05[ad].isna().sum()}")

    if ortak_05.any():
        mae = baseline_05.loc[ortak_05, ad + " hata"].mean()
        print(f"{ad} — ön MAE: {mae:.4f} atım/dk")
    else:
        print(f"{ad} — ön MAE: ___")

print("\nHam yöntemde en yüksek hatalı 5 pencere:")
print(baseline_05.nlargest(5, "Ham tepe hata").to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
n_05 = pencere_sayisi_05 * pencere_05
x_05 = ppg_05[:n_05].reshape(pencere_sayisi_05, pencere_05)

guc_05 = np.mean(
    (x_05 - x_05.mean(axis=1, keepdims=True)) ** 2,
    axis=1, keepdims=True
)
if np.any(guc_05 <= 0):
    raise ValueError("En az bir pencerenin sinyal gücü sıfır.")

referans_05 = baseline_05["EKG nabzı"].to_numpy()
deney_satirlari_05 = []

for tekrar in range(5):
    rng_05 = np.random.default_rng(4051 + tekrar)
    z_05 = rng_05.normal(size=x_05.shape)
    z_05 -= z_05.mean(axis=1, keepdims=True)
    z_05 /= np.sqrt(np.mean(z_05 ** 2, axis=1, keepdims=True))

    for snr_db in [20, 10, 0]:
        gurultu_05 = z_05 * np.sqrt(
            guc_05 / (10 ** (snr_db / 10))
        )
        x_gurultulu_05 = (x_05 + gurultu_05).ravel()

        sinyaller_deney_05 = {
            "Ham tepe": x_gurultulu_05,
            "Hareketli ortalama": uniform_filter1d(
                x_gurultulu_05, size=ma_ornek_05, mode="reflect"
            ),
            "Butterworth": sosfiltfilt(sos_05, x_gurultulu_05)
        }

        tahminler_05 = {}
        for ad, sinyal in sinyaller_deney_05.items():
            nabizlar = np.full(pencere_sayisi_05, np.nan)

            for i in range(pencere_sayisi_05):
                a = i * pencere_05
                b = a + pencere_05
                pk, _ = find_peaks(
                    sinyal[a:b],
                    distance=int(np.ceil(0.30 * fs_05)),
                    prominence=0.05
                )
                if len(pk) >= 2:
                    nabizlar[i] = 60 / np.mean(np.diff(pk) / fs_05)

            tahminler_05[ad] = nabizlar

        ortak = np.isfinite(referans_05)
        for nabizlar in tahminler_05.values():
            ortak = ortak & np.isfinite(nabizlar)

        for ad, nabizlar in tahminler_05.items():
            mae = np.nan
            if ortak.any():
                mae = np.mean(abs(
                    nabizlar[ortak] - referans_05[ortak]
                ))

            deney_satirlari_05.append({
                "Kayıt": "bidmc_05",
                "SNR (dB)": snr_db,
                "Tekrar": tekrar + 1,
                "Yöntem": ad,
                "MAE": mae,
                "Ortak pencere": int(ortak.sum()),
                "Hesaplanamayan": int(np.isnan(nabizlar).sum())
            })

deney_sonuclari_05 = pd.DataFrame(deney_satirlari_05)

ozet_05 = deney_sonuclari_05.groupby(
    ["SNR (dB)", "Yöntem"]
).agg(
    Ortalama_MAE=("MAE", "mean"),
    MAE_std=("MAE", "std"),
    En_az_ortak_pencere=("Ortak pencere", "min"),
    En_cok_ortak_pencere=("Ortak pencere", "max"),
    Toplam_hesaplanamayan=("Hesaplanamayan", "sum")
).reset_index().sort_values(
    ["SNR (dB)", "Yöntem"], ascending=[False, True]
)

deney_sonuclari_05.to_csv("bidmc_05_gurultu_tekrarlari.csv", index=False)
ozet_05.to_csv("bidmc_05_SNR_ozet.csv", index=False)

print(ozet_05.to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
sabit_maske_05 = ortak_05.to_numpy().copy()
sabit_satirlar_05 = []

for tekrar in range(5):
    rng_sabit_05 = np.random.default_rng(4051 + tekrar)
    z = rng_sabit_05.normal(size=x_05.shape)
    z -= z.mean(axis=1, keepdims=True)
    z /= np.sqrt(np.mean(z ** 2, axis=1, keepdims=True))

    for snr_db in [20, 10, 0]:
        gurultulu = (
            x_05 + z * np.sqrt(guc_05 / (10 ** (snr_db / 10)))
        ).ravel()

        sinyaller = {
            "Ham tepe": gurultulu,
            "Hareketli ortalama": uniform_filter1d(
                gurultulu, size=ma_ornek_05, mode="reflect"
            ),
            "Butterworth": sosfiltfilt(sos_05, gurultulu)
        }

        for ad, sinyal in sinyaller.items():
            for i in range(pencere_sayisi_05):
                a = i * pencere_05
                b = a + pencere_05
                pk, _ = find_peaks(
                    sinyal[a:b],
                    distance=int(np.ceil(0.30 * fs_05)),
                    prominence=0.05
                )

                tahmin = np.nan
                if len(pk) >= 2:
                    tahmin = 60 / np.mean(np.diff(pk) / fs_05)

                sabit_satirlar_05.append({
                    "SNR (dB)": snr_db,
                    "Tekrar": tekrar + 1,
                    "Yöntem": ad,
                    "Başlangıç (s)": a / fs_05,
                    "Sabit kümeye dahil": bool(sabit_maske_05[i]),
                    "PPG nabzı": tahmin,
                    "Mutlak hata": abs(tahmin - referans_05[i])
                })

pencere_deney_05 = pd.DataFrame(sabit_satirlar_05)
secilen_05 = pencere_deney_05[
    pencere_deney_05["Sabit kümeye dahil"]
]

sabit_tekrar_05 = secilen_05.groupby(
    ["SNR (dB)", "Tekrar", "Yöntem"]
).agg(
    MAE=("Mutlak hata", "mean"),
    Hesaplanan=("Mutlak hata", "count")
).reset_index()

sabit_tekrar_05.loc[
    sabit_tekrar_05["Hesaplanan"] != int(sabit_maske_05.sum()), "MAE"
] = np.nan

ozet_sabit_05 = sabit_tekrar_05.groupby(
    ["SNR (dB)", "Yöntem"]
).agg(
    Ortalama_MAE=("MAE", "mean"),
    MAE_std=("MAE", "std"),
    Tam_tekrar_sayisi=("MAE", "count"),
    En_az_hesaplanan=("Hesaplanan", "min")
).reset_index().sort_values(
    ["SNR (dB)", "Yöntem"], ascending=[False, True]
)

pencere_deney_05.to_csv("bidmc_05_pencere_deneyleri.csv", index=False)
sabit_tekrar_05.to_csv("bidmc_05_sabit_tekrarlar.csv", index=False)
ozet_sabit_05.to_csv("bidmc_05_SNR_sabit_ozet.csv", index=False)

print("Sabit değerlendirme pencere sayısı:", sabit_maske_05.sum())
print(ozet_sabit_05.to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
from scipy.signal import butter, sosfiltfilt, find_peaks
from scipy.ndimage import uniform_filter1d

pencere_06 = int(round(10 * fs_06))
pencere_sayisi_06 = len(ppg_06) // pencere_06

ma_ornek_06 = int(round(0.10 * fs_06))
if ma_ornek_06 % 2 == 0:
    ma_ornek_06 += 1

sos_06 = butter(
    4, [0.5, 8], btype="bandpass", fs=fs_06, output="sos"
)

sinyaller_06 = {
    "Ham tepe": ppg_06,
    "Hareketli ortalama": uniform_filter1d(
        ppg_06, size=ma_ornek_06, mode="reflect"
    ),
    "Butterworth": sosfiltfilt(sos_06, ppg_06)
}

satirlar_06 = []

for i in range(pencere_sayisi_06):
    a = i * pencere_06
    b = a + pencere_06

    r_pk, _ = find_peaks(
        ekg_06[a:b],
        distance=int(np.ceil(0.30 * fs_06)),
        prominence=0.30
    )

    ekg_nabiz = np.nan
    if len(r_pk) >= 2:
        ekg_nabiz = 60 / np.mean(np.diff(r_pk) / fs_06)

    satir = {
        "Başlangıç (s)": a / fs_06,
        "EKG V adayı sayısı": len(r_pk),
        "EKG nabzı": ekg_nabiz
    }

    for ad, sinyal in sinyaller_06.items():
        pk, _ = find_peaks(
            sinyal[a:b],
            distance=int(np.ceil(0.30 * fs_06)),
            prominence=0.05
        )

        nabiz = np.nan
        if len(pk) >= 2:
            nabiz = 60 / np.mean(np.diff(pk) / fs_06)

        satir[ad] = nabiz
        satir[ad + " tepe sayısı"] = len(pk)

    satirlar_06.append(satir)

baseline_06 = pd.DataFrame(satirlar_06)
adlar_06 = list(sinyaller_06)
nabiz_sutunlari_06 = ["EKG nabzı"] + adlar_06
ortak_06 = baseline_06[nabiz_sutunlari_06].notna().all(axis=1)

print(f"Ortak hesaplanabilen pencere: "
      f"{ortak_06.sum()} / {pencere_sayisi_06}")
print("EKG hesaplanamayan pencere:",
      baseline_06["EKG nabzı"].isna().sum())

for ad in adlar_06:
    baseline_06[ad + " hata"] = abs(
        baseline_06[ad] - baseline_06["EKG nabzı"]
    )
    print(f"{ad} — hesaplanamayan pencere: "
          f"{baseline_06[ad].isna().sum()}")

    if ortak_06.any():
        mae = baseline_06.loc[ortak_06, ad + " hata"].mean()
        print(f"{ad} — ön MAE: {mae:.4f} atım/dk")
    else:
        print(f"{ad} — ön MAE: ___")

print("\nHam yöntemde en yüksek hatalı 5 pencere:")
print(baseline_06.nlargest(5, "Ham tepe hata").to_string(
    index=False, float_format="%.4f", na_rep="___"
))

eksik_06 = ~baseline_06[nabiz_sutunlari_06].notna().all(axis=1)
print("\nEn az bir nabız sonucunun eksik olduğu pencereler:")
print(baseline_06.loc[eksik_06].to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
n_06 = pencere_sayisi_06 * pencere_06
x_06 = ppg_06[:n_06].reshape(pencere_sayisi_06, pencere_06)

guc_06 = np.mean(
    (x_06 - x_06.mean(axis=1, keepdims=True)) ** 2,
    axis=1, keepdims=True
)
if np.any(guc_06 <= 0):
    raise ValueError("En az bir pencerenin sinyal gücü sıfır.")

referans_06 = baseline_06["EKG nabzı"].to_numpy()
sabit_maske_06 = ortak_06.to_numpy().copy()
deney_satirlari_06 = []

for tekrar in range(5):
    rng_06 = np.random.default_rng(4061 + tekrar)
    z = rng_06.normal(size=x_06.shape)
    z -= z.mean(axis=1, keepdims=True)
    z /= np.sqrt(np.mean(z ** 2, axis=1, keepdims=True))

    for snr_db in [20, 10, 0]:
        gurultulu = (
            x_06 + z * np.sqrt(guc_06 / (10 ** (snr_db / 10)))
        ).ravel()

        sinyaller = {
            "Ham tepe": gurultulu,
            "Hareketli ortalama": uniform_filter1d(
                gurultulu, size=ma_ornek_06, mode="reflect"
            ),
            "Butterworth": sosfiltfilt(sos_06, gurultulu)
        }

        for ad, sinyal in sinyaller.items():
            for i in range(pencere_sayisi_06):
                a = i * pencere_06
                b = a + pencere_06
                pk, _ = find_peaks(
                    sinyal[a:b],
                    distance=int(np.ceil(0.30 * fs_06)),
                    prominence=0.05
                )

                tahmin = np.nan
                if len(pk) >= 2:
                    tahmin = 60 / np.mean(np.diff(pk) / fs_06)

                deney_satirlari_06.append({
                    "Kayıt": "bidmc_06",
                    "EKG referansı": "V, yukarı yön, prominence=0.30",
                    "SNR (dB)": snr_db,
                    "Tekrar": tekrar + 1,
                    "Yöntem": ad,
                    "Başlangıç (s)": a / fs_06,
                    "Sabit kümeye dahil": bool(sabit_maske_06[i]),
                    "PPG nabzı": tahmin,
                    "Mutlak hata": abs(tahmin - referans_06[i])
                })

pencere_deney_06 = pd.DataFrame(deney_satirlari_06)
secilen_06 = pencere_deney_06[
    pencere_deney_06["Sabit kümeye dahil"]
]

sabit_tekrar_06 = secilen_06.groupby(
    ["SNR (dB)", "Tekrar", "Yöntem"]
).agg(
    MAE=("Mutlak hata", "mean"),
    Hesaplanan=("Mutlak hata", "count")
).reset_index()

sabit_tekrar_06.loc[
    sabit_tekrar_06["Hesaplanan"] != int(sabit_maske_06.sum()), "MAE"
] = np.nan

ozet_sabit_06 = sabit_tekrar_06.groupby(
    ["SNR (dB)", "Yöntem"]
).agg(
    Ortalama_MAE=("MAE", "mean"),
    MAE_std=("MAE", "std"),
    Tam_tekrar_sayisi=("MAE", "count"),
    En_az_hesaplanan=("Hesaplanan", "min")
).reset_index().sort_values(
    ["SNR (dB)", "Yöntem"], ascending=[False, True]
)

pencere_deney_06.to_csv("bidmc_06_pencere_deneyleri.csv", index=False)
sabit_tekrar_06.to_csv("bidmc_06_sabit_tekrarlar.csv", index=False)
ozet_sabit_06.to_csv("bidmc_06_SNR_sabit_ozet.csv", index=False)

print("Sabit değerlendirme pencere sayısı:", sabit_maske_06.sum())
print(ozet_sabit_06.to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
from scipy.signal import butter, sosfiltfilt, find_peaks
from scipy.ndimage import uniform_filter1d

pencere_08 = int(round(10 * fs_08))
pencere_sayisi_08 = len(ppg_08) // pencere_08

ma_ornek_08 = int(round(0.10 * fs_08))
if ma_ornek_08 % 2 == 0:
    ma_ornek_08 += 1

sos_08 = butter(
    4, [0.5, 8], btype="bandpass", fs=fs_08, output="sos"
)

sinyaller_08 = {
    "Ham tepe": ppg_08,
    "Hareketli ortalama": uniform_filter1d(
        ppg_08, size=ma_ornek_08, mode="reflect"
    ),
    "Butterworth": sosfiltfilt(sos_08, ppg_08)
}

satirlar_08 = []

for i in range(pencere_sayisi_08):
    a = i * pencere_08
    b = a + pencere_08

    r_pk, _ = find_peaks(
        ekg_08[a:b],
        distance=int(np.ceil(0.30 * fs_08)),
        prominence=0.40
    )

    ekg_nabiz = np.nan
    if len(r_pk) >= 2:
        ekg_nabiz = 60 / np.mean(np.diff(r_pk) / fs_08)

    satir = {
        "Başlangıç (s)": a / fs_08,
        "EKG II adayı sayısı": len(r_pk),
        "EKG nabzı": ekg_nabiz
    }

    for ad, sinyal in sinyaller_08.items():
        pk, _ = find_peaks(
            sinyal[a:b],
            distance=int(np.ceil(0.30 * fs_08)),
            prominence=0.05
        )

        nabiz = np.nan
        if len(pk) >= 2:
            nabiz = 60 / np.mean(np.diff(pk) / fs_08)

        satir[ad] = nabiz
        satir[ad + " tepe sayısı"] = len(pk)

    satirlar_08.append(satir)

baseline_08 = pd.DataFrame(satirlar_08)
adlar_08 = list(sinyaller_08)
nabiz_sutunlari_08 = ["EKG nabzı"] + adlar_08
ortak_08 = baseline_08[nabiz_sutunlari_08].notna().all(axis=1)

print(f"Ortak hesaplanabilen pencere: "
      f"{ortak_08.sum()} / {pencere_sayisi_08}")
print("EKG hesaplanamayan pencere:",
      baseline_08["EKG nabzı"].isna().sum())

for ad in adlar_08:
    baseline_08[ad + " hata"] = abs(
        baseline_08[ad] - baseline_08["EKG nabzı"]
    )
    print(f"{ad} — hesaplanamayan pencere: "
          f"{baseline_08[ad].isna().sum()}")

    if ortak_08.any():
        mae = baseline_08.loc[ortak_08, ad + " hata"].mean()
        print(f"{ad} — ön MAE: {mae:.4f} atım/dk")
    else:
        print(f"{ad} — ön MAE: ___")

print("\nHam yöntemde en yüksek hatalı 5 pencere:")
print(baseline_08.nlargest(5, "Ham tepe hata").to_string(
    index=False, float_format="%.4f", na_rep="___"
))

print("\nEn az bir nabız sonucunun eksik olduğu pencereler:")
print(baseline_08.loc[~ortak_08].to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
from scipy.signal import butter, sosfiltfilt, find_peaks
from scipy.ndimage import uniform_filter1d

pencere_09 = int(round(10 * fs_09))
pencere_sayisi_09 = len(ppg_09) // pencere_09

ma_ornek_09 = int(round(0.10 * fs_09))
if ma_ornek_09 % 2 == 0:
    ma_ornek_09 += 1

sos_09 = butter(
    4, [0.5, 8], btype="bandpass", fs=fs_09, output="sos"
)

sinyaller_09 = {
    "Ham tepe": ppg_09,
    "Hareketli ortalama": uniform_filter1d(
        ppg_09, size=ma_ornek_09, mode="reflect"
    ),
    "Butterworth": sosfiltfilt(sos_09, ppg_09)
}

satirlar_09 = []

for i in range(pencere_sayisi_09):
    a = i * pencere_09
    b = a + pencere_09

    r_pk, _ = find_peaks(
        ekg_09[a:b],
        distance=int(np.ceil(0.30 * fs_09)),
        prominence=0.35
    )

    ekg_nabiz = np.nan
    if len(r_pk) >= 2:
        ekg_nabiz = 60 / np.mean(np.diff(r_pk) / fs_09)

    satir = {
        "Başlangıç (s)": a / fs_09,
        "EKG II adayı sayısı": len(r_pk),
        "EKG nabzı": ekg_nabiz
    }

    for ad, sinyal in sinyaller_09.items():
        pk, _ = find_peaks(
            sinyal[a:b],
            distance=int(np.ceil(0.30 * fs_09)),
            prominence=0.05
        )

        nabiz = np.nan
        if len(pk) >= 2:
            nabiz = 60 / np.mean(np.diff(pk) / fs_09)

        satir[ad] = nabiz
        satir[ad + " tepe sayısı"] = len(pk)

    satirlar_09.append(satir)

baseline_09 = pd.DataFrame(satirlar_09)
adlar_09 = list(sinyaller_09)
nabiz_sutunlari_09 = ["EKG nabzı"] + adlar_09
ortak_09 = baseline_09[nabiz_sutunlari_09].notna().all(axis=1)

print(f"Ortak hesaplanabilen pencere: "
      f"{ortak_09.sum()} / {pencere_sayisi_09}")
print("EKG hesaplanamayan pencere:",
      baseline_09["EKG nabzı"].isna().sum())

for ad in adlar_09:
    baseline_09[ad + " hata"] = abs(
        baseline_09[ad] - baseline_09["EKG nabzı"]
    )
    print(f"{ad} — hesaplanamayan pencere: "
          f"{baseline_09[ad].isna().sum()}")

    if ortak_09.any():
        mae = baseline_09.loc[ortak_09, ad + " hata"].mean()
        print(f"{ad} — ön MAE: {mae:.4f} atım/dk")
    else:
        print(f"{ad} — ön MAE: ___")

print("\nHam yöntemde en yüksek hatalı 5 pencere:")
print(baseline_09.nlargest(5, "Ham tepe hata").to_string(
    index=False, float_format="%.4f", na_rep="___"
))

print("\nEn az bir nabız sonucunun eksik olduğu pencereler:")
print(baseline_09.loc[~ortak_09].to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
from scipy.signal import butter, sosfiltfilt, find_peaks
from scipy.ndimage import uniform_filter1d

pencere_10 = int(round(10 * fs_10))
pencere_sayisi_10 = len(ppg_10) // pencere_10

ma_ornek_10 = int(round(0.10 * fs_10))
if ma_ornek_10 % 2 == 0:
    ma_ornek_10 += 1

sos_10 = butter(
    4, [0.5, 8], btype="bandpass", fs=fs_10, output="sos"
)

sinyaller_10 = {
    "Ham tepe": ppg_10,
    "Hareketli ortalama": uniform_filter1d(
        ppg_10, size=ma_ornek_10, mode="reflect"
    ),
    "Butterworth": sosfiltfilt(sos_10, ppg_10)
}

satirlar_10 = []

for i in range(pencere_sayisi_10):
    a = i * pencere_10
    b = a + pencere_10

    v_pk, _ = find_peaks(
        -ekg_10[a:b],
        distance=int(np.ceil(0.30 * fs_10)),
        prominence=0.30
    )

    ekg_nabiz = np.nan
    if len(v_pk) >= 2:
        ekg_nabiz = 60 / np.mean(np.diff(v_pk) / fs_10)

    satir = {
        "Başlangıç (s)": a / fs_10,
        "EKG V adayı sayısı": len(v_pk),
        "EKG nabzı": ekg_nabiz
    }

    for ad, sinyal in sinyaller_10.items():
        pk, _ = find_peaks(
            sinyal[a:b],
            distance=int(np.ceil(0.30 * fs_10)),
            prominence=0.05
        )

        nabiz = np.nan
        if len(pk) >= 2:
            nabiz = 60 / np.mean(np.diff(pk) / fs_10)

        satir[ad] = nabiz
        satir[ad + " tepe sayısı"] = len(pk)

    satirlar_10.append(satir)

baseline_10 = pd.DataFrame(satirlar_10)
adlar_10 = list(sinyaller_10)
nabiz_sutunlari_10 = ["EKG nabzı"] + adlar_10
ortak_10 = baseline_10[nabiz_sutunlari_10].notna().all(axis=1)

print(f"Ortak hesaplanabilen pencere: "
      f"{ortak_10.sum()} / {pencere_sayisi_10}")
print("EKG hesaplanamayan pencere:",
      baseline_10["EKG nabzı"].isna().sum())

for ad in adlar_10:
    baseline_10[ad + " hata"] = abs(
        baseline_10[ad] - baseline_10["EKG nabzı"]
    )
    print(f"{ad} — hesaplanamayan pencere: "
          f"{baseline_10[ad].isna().sum()}")

    if ortak_10.any():
        mae = baseline_10.loc[ortak_10, ad + " hata"].mean()
        print(f"{ad} — ön MAE: {mae:.4f} atım/dk")
    else:
        print(f"{ad} — ön MAE: ___")

print("\nHam yöntemde en yüksek hatalı 5 pencere:")
print(baseline_10.nlargest(5, "Ham tepe hata").to_string(
    index=False, float_format="%.4f", na_rep="___"
))

print("\nEn az bir nabız sonucunun eksik olduğu pencereler:")
print(baseline_10.loc[~ortak_10].to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
n_10 = pencere_sayisi_10 * pencere_10
x_10 = ppg_10[:n_10].reshape(pencere_sayisi_10, pencere_10)

guc_10 = np.mean(
    (x_10 - x_10.mean(axis=1, keepdims=True)) ** 2,
    axis=1, keepdims=True
)
if np.any(guc_10 <= 0):
    raise ValueError("En az bir pencerenin sinyal gücü sıfır.")

referans_10 = baseline_10["EKG nabzı"].to_numpy()
sabit_maske_10 = ortak_10.to_numpy().copy()
deney_satirlari_10 = []

for tekrar in range(5):
    rng_10 = np.random.default_rng(4101 + tekrar)
    z = rng_10.normal(size=x_10.shape)
    z -= z.mean(axis=1, keepdims=True)
    z /= np.sqrt(np.mean(z ** 2, axis=1, keepdims=True))

    for snr_db in [20, 10, 0]:
        gurultulu = (
            x_10 + z * np.sqrt(guc_10 / (10 ** (snr_db / 10)))
        ).ravel()

        sinyaller = {
            "Ham tepe": gurultulu,
            "Hareketli ortalama": uniform_filter1d(
                gurultulu, size=ma_ornek_10, mode="reflect"
            ),
            "Butterworth": sosfiltfilt(sos_10, gurultulu)
        }

        for ad, sinyal in sinyaller.items():
            for i in range(pencere_sayisi_10):
                a = i * pencere_10
                b = a + pencere_10
                pk, _ = find_peaks(
                    sinyal[a:b],
                    distance=int(np.ceil(0.30 * fs_10)),
                    prominence=0.05
                )

                tahmin = np.nan
                if len(pk) >= 2:
                    tahmin = 60 / np.mean(np.diff(pk) / fs_10)

                deney_satirlari_10.append({
                    "Kayıt": "bidmc_10",
                    "EKG referansı": "V, aşağı yön, prominence=0.30 — geçici",
                    "SNR (dB)": snr_db,
                    "Tekrar": tekrar + 1,
                    "Yöntem": ad,
                    "Başlangıç (s)": a / fs_10,
                    "Sabit kümeye dahil": bool(sabit_maske_10[i]),
                    "PPG nabzı": tahmin,
                    "Mutlak hata": abs(tahmin - referans_10[i])
                })

pencere_deney_10 = pd.DataFrame(deney_satirlari_10)
secilen_10 = pencere_deney_10[
    pencere_deney_10["Sabit kümeye dahil"]
]

sabit_tekrar_10 = secilen_10.groupby(
    ["SNR (dB)", "Tekrar", "Yöntem"]
).agg(
    MAE=("Mutlak hata", "mean"),
    Hesaplanan=("Mutlak hata", "count")
).reset_index()

sabit_tekrar_10.loc[
    sabit_tekrar_10["Hesaplanan"] != int(sabit_maske_10.sum()), "MAE"
] = np.nan

ozet_sabit_10 = sabit_tekrar_10.groupby(
    ["SNR (dB)", "Yöntem"]
).agg(
    Ortalama_MAE=("MAE", "mean"),
    MAE_std=("MAE", "std"),
    Tam_tekrar_sayisi=("MAE", "count"),
    En_az_hesaplanan=("Hesaplanan", "min")
).reset_index().sort_values(
    ["SNR (dB)", "Yöntem"], ascending=[False, True]
)

pencere_deney_10.to_csv("bidmc_10_pencere_deneyleri.csv", index=False)
sabit_tekrar_10.to_csv("bidmc_10_sabit_tekrarlar.csv", index=False)
ozet_sabit_10.to_csv("bidmc_10_SNR_sabit_ozet.csv", index=False)

print("Sabit değerlendirme pencere sayısı:", sabit_maske_10.sum())
print(ozet_sabit_10.to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
import re
import pandas as pd
from io import BytesIO

birlesen = []

for i in range(1, 11):
    kayit = f"bidmc_{i:02d}"
    tur = "SNR_ozet" if i <= 4 else "SNR_sabit_ozet"
    desen = rf"{kayit}_{tur}(?: \(\d+\))?\.csv"

    adaylar = sorted(
        ad for ad in yuklenen_ozetler
        if re.fullmatch(desen, ad)
    )
    if not adaylar:
        raise ValueError(f"Eksik özet: {kayit}")

    tablolar = [
        pd.read_csv(BytesIO(yuklenen_ozetler[ad]))
        for ad in adaylar
    ]
    tablo = tablolar[0]

    if any(not tablo.equals(diger) for diger in tablolar[1:]):
        raise ValueError(f"{kayit}: kopya dosyaların içerikleri farklı.")

    if (
        len(tablo) != 9
        or tablo.duplicated(["SNR (dB)", "Yöntem"]).any()
        or set(tablo["SNR (dB)"]) != {0, 10, 20}
        or set(tablo["Yöntem"]) != {
            "Ham tepe", "Hareketli ortalama", "Butterworth"
        }
        or tablo["Ortalama_MAE"].isna().any()
    ):
        raise ValueError(f"{kayit}: özet tablosunu kontrol etmeliyiz.")

    parca = tablo[["SNR (dB)", "Yöntem", "Ortalama_MAE"]].copy()
    parca["Kayıt"] = kayit
    birlesen.append(parca)
    print("Seçildi:", adaylar[0])

kayit_ozetleri = pd.concat(birlesen, ignore_index=True)

genel_ozet = kayit_ozetleri.groupby(
    ["SNR (dB)", "Yöntem"]
).agg(
    Kayitlar_ortalama_MAE=("Ortalama_MAE", "mean"),
    Kayitlar_arasi_std=("Ortalama_MAE", "std"),
    Kayit_sayisi=("Ortalama_MAE", "count")
).reset_index().sort_values(
    ["SNR (dB)", "Yöntem"], ascending=[False, True]
)

kayit_ozetleri.to_csv("bidmc_01_10_kayit_ozetleri_on.csv", index=False)
genel_ozet.to_csv("bidmc_01_10_genel_ozet_on.csv", index=False)

print("\nOn kaydın ortak ön özeti:")
print(genel_ozet.to_string(
    index=False, float_format="%.4f", na_rep="___"
))
```

```python
import matplotlib.pyplot as plt

fig, eksenler = plt.subplots(1, 2, figsize=(13, 5))

for ad, renk in zip(
    ["Ham tepe", "Hareketli ortalama", "Butterworth"],
    ["tab:red", "tab:green", "tab:blue"]
):
    veri = genel_ozet[
        genel_ozet["Yöntem"] == ad
    ].sort_values("SNR (dB)", ascending=False)

    for panel in ([0] if ad == "Ham tepe" else [0, 1]):
        eksenler[panel].errorbar(
            veri["SNR (dB)"],
            veri["Kayitlar_ortalama_MAE"],
            yerr=veri["Kayitlar_arasi_std"],
            marker="o", capsize=4, color=renk, label=ad
        )

eksenler[0].set_title("Üç yöntemin karşılaştırması")
eksenler[1].set_title("Filtre yöntemlerinin ayrıntısı")

for eksen in eksenler:
    eksen.set_xlabel("SNR (dB) — sağa doğru gürültü artar")
    eksen.set_ylabel("Kayıtlar üzerinden ortalama MAE (atım/dk)")
    eksen.set_xticks([20, 10, 0])
    eksen.set_xlim(21, -1)
    eksen.grid(True, alpha=0.3)
    eksen.legend()

fig.suptitle("bidmc_01–10 — Beyaz gürültü altında nabız tahmin hatası")
fig.text(
    0.5, 0.02,
    "Her kayda eşit ağırlık; kayıt başına 5 gürültü tekrarı.\n"
    "Çubuklar: kayıtlar arası ±1 standart sapma; güven aralığı değildir.\n"
    "Mevcut PPG bozulmaları ve EKG referans belirsizlikleri dahil; ön sonuç.",
    ha="center", fontsize=9
)
fig.tight_layout(rect=[0, 0.14, 1, 0.93])
fig.savefig("bidmc_01_10_genel_MAE_SNR_on.png",
            dpi=300, bbox_inches="tight")
plt.show()
```
